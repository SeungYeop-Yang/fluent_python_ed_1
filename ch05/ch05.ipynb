```python
def factorial(n):
    """returns n!"""

    return 1 if n < 2 else n * factorial(n - 1)
```

```python
factorial(42)
```

```python
type(factorial)
```

```python
fact = factorial
fact  
```

```python
fact(5)
```

```python
map(factorial, range(11))
```

```python
fruits = ['strawberry', 'fig', 'apple', 'cherry', 'raspberry', 'banana']
sorted(fruits, key=len)
```

```python
print(list(reversed("abcd")))
```

```python
def reverse(word):
    return word[::-1]   
reverse('abcd')
```
